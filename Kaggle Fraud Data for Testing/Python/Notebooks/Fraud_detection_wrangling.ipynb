```python
# Dropping Duplicates
FD = FD.drop_duplicates()
```

```python
# Copy FD to avoid overwriting
FD1 = FD.copy()
```

# Make it a proper fintech dataset

```python
# Add unique transaction ID
FD1['transaction_id'] = range(1, len(FD)+1)
```

```python
# Simulate customer IDs
FD1['customer_id'] = np.random.randint(1000, 2000, size=len(FD))
```

```python
# Rename columns
FD1.rename(columns={
    'Time': 'transaction_time',
    'Amount': 'transaction_amount',
    'Class': 'is_fraud'
}, inplace=True)
```

```python
# Reorder columns
cols = ['transaction_id', 'customer_id', 'transaction_time', 
        'transaction_amount', 'is_fraud'] + \
       [col for col in FD1.columns if col.startswith('V')]
FD1 = FD1[cols]
```

```python
fraud_ratio = FD1['is_fraud'].value_counts(normalize=True)
print(fraud_ratio)
```
